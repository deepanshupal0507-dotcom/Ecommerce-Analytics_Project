```python
import pandas as pd
import os

data_path = "Data/Raw"

orders = pd.read_csv(os.path.join(data_path, "olist_orders_dataset.csv"))
customers = pd.read_csv(os.path.join(data_path, "olist_customers_dataset.csv"))
order_items = pd.read_csv(os.path.join(data_path, "olist_order_items_dataset.csv"))
payments = pd.read_csv(os.path.join(data_path, "olist_order_payments_dataset.csv"))
reviews = pd.read_csv(os.path.join(data_path, "olist_order_reviews_dataset.csv"))
products = pd.read_csv(os.path.join(data_path, "olist_products_dataset.csv"))
sellers = pd.read_csv(os.path.join(data_path, "olist_sellers_dataset.csv"))
geolocation = pd.read_csv(os.path.join(data_path, "olist_geolocation_dataset.csv"))
category_translation = pd.read_csv(os.path.join(data_path, "product_category_name_translation.csv"))

print("All datasets loaded successfully!")

print("Orders:", orders.shape)
print("Customers:", customers.shape)
print("Order Items:", order_items.shape)
print("Payments:", payments.shape)
print("Reviews:", reviews.shape)
print("Products:", products.shape)
print("Sellers:", sellers.shape)
print("Geolocation:", geolocation.shape)
print("Category Translation:", category_translation.shape)
```

```python
datasets = {
    "Orders": orders,
    "Customers": customers,
    "Order Items": order_items,
    "Payments": payments,
    "Reviews": reviews,
    "Products": products,
    "Sellers": sellers,
    "Geolocation": geolocation,
    "Category Translation": category_translation
}

for name, df in datasets.items():
    print(f"\n{name}")
    print("Missing values:", df.isnull().sum().sum())
```

```python
for name, df in datasets.items():
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if not missing.empty:
        print(f"\n{name}")
        print(missing)
```

```python
print("Orders before cleaning:", orders.shape)

print("\nMissing values before cleaning:")
print(orders.isnull().sum())
```

```python
# Clean missing values in Orders dataset

orders['order_approved_at'] = orders['order_approved_at'].fillna(
    orders['order_purchase_timestamp']
)

orders['order_delivered_carrier_date'] = orders[
    'order_delivered_carrier_date'
].fillna('Not Available')

orders['order_delivered_customer_date'] = orders[
    'order_delivered_customer_date'
].fillna('Not Available')

print("Orders dataset cleaned successfully!")
print("\nMissing values after cleaning:")
print(orders.isnull().sum())
```

```python
# Clean Reviews dataset
reviews['review_comment_title'] = reviews[
    'review_comment_title'
].fillna('No Title')

reviews['review_comment_message'] = reviews[
    'review_comment_message'
].fillna('No Comment')

# Clean Products dataset
products['product_category_name'] = products[
    'product_category_name'
].fillna('Unknown')

products['product_name_lenght'] = products[
    'product_name_lenght'
].fillna(0)

products['product_description_lenght'] = products[
    'product_description_lenght'
].fillna(0)

products['product_photos_qty'] = products[
    'product_photos_qty'
].fillna(0)

print("Reviews and Products cleaned successfully!")

print("\nReviews missing values:")
print(reviews.isnull().sum())

print("\nProducts missing values:")
print(products.isnull().sum())
```

```python
# Fill missing product measurements with median values

measurement_columns = [
    'product_weight_g',
    'product_length_cm',
    'product_height_cm',
    'product_width_cm'
]

for col in measurement_columns:
    products[col] = products[col].fillna(products[col].median())

print("Product measurements cleaned successfully!")
print("\nRemaining missing values:")
print(products.isnull().sum())
```

```python
# Step 4: Basic E-commerce Data Analysis

import matplotlib.pyplot as plt
import seaborn as sns

# Convert purchase timestamp to datetime
orders['order_purchase_timestamp'] = pd.to_datetime(
    orders['order_purchase_timestamp'],
    errors='coerce'
)

# Extract year and month
orders['year_month'] = orders[
    'order_purchase_timestamp'
].dt.to_period('M').astype(str)

# Total orders
print("Total Orders:", orders['order_id'].nunique())

# Order status analysis
print("\nOrders by Status:")
print(orders['order_status'].value_counts())

# Monthly order trend
monthly_orders = orders.groupby('year_month')[
    'order_id'
].nunique()

plt.figure(figsize=(12, 5))
monthly_orders.plot(kind='line', marker='o')
plt.title('Monthly E-commerce Orders Trend')
plt.xlabel('Year-Month')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
print("First order date:", orders["order_purchase_timestamp"].min())
print("Last order date:", orders["order_purchase_timestamp"].max())

print("\nMonthly orders:")
print(
    orders.groupby(
        orders["order_purchase_timestamp"].dt.to_period("M")
    ).size().tail(10)
)
```

```python
# Revenue Analysis

import pandas as pd
import matplotlib.pyplot as plt

# Convert date column to datetime
orders['order_purchase_timestamp'] = pd.to_datetime(
    orders['order_purchase_timestamp']
)

# Convert payment date into useful sales data
print("Orders dataset:", orders.shape)
print("Payments dataset:", payments.shape)

print("\nPayment columns:")
print(payments.columns.tolist())

print("\nPayment methods:")
print(payments['payment_type'].value_counts())
```

```python
# Total Revenue Analysis

print("Total Revenue:", payments['payment_value'].sum())

print("\nAverage Payment Value:",
      payments['payment_value'].mean())

print("\nTotal Payments:",
      len(payments))

print("\nPayment Value Statistics:")
print(payments['payment_value'].describe())
```

```python
# Monthly Revenue Analysis

payments['payment_value'] = pd.to_numeric(
    payments['payment_value'], errors='coerce'
)

monthly_revenue = payments.groupby(
    payments['order_id'].map(
        orders.set_index('order_id')['order_purchase_timestamp']
    ).pipe(pd.to_datetime).dt.to_period('M')
)['payment_value'].sum()

monthly_revenue.index = monthly_revenue.index.astype(str)

print("Monthly Revenue:")
print(monthly_revenue)

plt.figure(figsize=(12, 5))
monthly_revenue.plot(kind='line', marker='o')
plt.title('Monthly E-commerce Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Payment Value')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Top 10 Product Categories by Revenue

# Merge order items with product details
items_products = order_items.merge(
    products[['product_id', 'product_category_name']],
    on='product_id',
    how='left'
)

# Merge with payment information
items_payments = items_products.merge(
    payments[['order_id', 'payment_value']],
    on='order_id',
    how='left'
)

# Calculate revenue by category
category_revenue = (
    items_payments.groupby('product_category_name')['payment_value']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 Product Categories:")
print(category_revenue)

# Plot the results
plt.figure(figsize=(12, 6))
category_revenue.sort_values().plot(kind='barh')
plt.title('Top 10 Product Categories by Payment Value')
plt.xlabel('Payment Value')
plt.ylabel('Product Category')
plt.tight_layout()
plt.show()
```

```python
# Customer Analysis

print("Total Unique Customers:", customers['customer_id'].nunique())

print("\nCustomer Dataset Shape:")
print(customers.shape)

print("\nTop 10 Customer States:")
print(customers['customer_state'].value_counts().head(10))

# Visualize top 10 customer states
import matplotlib.pyplot as plt

customers['customer_state'].value_counts().head(10).plot(
    kind='bar',
    figsize=(10, 5),
    title='Top 10 States by Number of Customers'
)

plt.xlabel('State')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Review Analysis

print("Total Reviews:", len(reviews))

print("\nReview Score Distribution:")
print(reviews['review_score'].value_counts().sort_index())

# Plot review score distribution
import matplotlib.pyplot as plt

reviews['review_score'].value_counts().sort_index().plot(
    kind='bar',
    figsize=(8, 5),
    title='Distribution of Customer Review Scores'
)

plt.xlabel('Review Score (1 to 5)')
plt.ylabel('Number of Reviews')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
